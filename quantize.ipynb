```python
import pandas as pd
```

```python
file_path = "TabulatedData/values_all_layers.csv"

# Step 1: Read only header to get column names
columns = pd.read_csv(file_path, nrows=0).columns

# Step 2: Define dtype mapping
dtype_map = {
    col: "float16" for col in columns if col != "name"
}
dtype_map["name"] = "string"   # pandas string dtype (more efficient than object)

# Step 3: Load dataset with predefined dtypes
df = pd.read_csv(
    file_path,
    dtype=dtype_map,
    low_memory=True,     # reduces peak memory usage
    memory_map=True      # faster file access for large files
)

print(df.info())
```

```python
df.to_parquet("values_all_layers_16bit.parquet", engine="fastparquet")
```
